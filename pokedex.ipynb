```python
# Importa a biblioteca requests, que permite fazer requisições para sites e APIs.
import requests

# Define uma função chamada buscar_pokemon que recebe um valor (nome ou número).
def buscar_pokemon(valor):

    # Cria a URL da PokeAPI usando o valor informado.
    url = f"https://pokeapi.co/api/v2/pokemon/{valor}"

    # Faz uma requisição GET para a URL e guarda a resposta.
    resposta = requests.get(url)

    # Verifica se a resposta foi bem-sucedida (código HTTP 200).
    if resposta.status_code == 200:

        # Converte a resposta da API (JSON) para um dicionário Python e retorna.
        return resposta.json()

    # Se o Pokémon não existir, retorna None.
    return None


# Exibe o título da Pokédex.
print("===== POKEDEX =====")

# Exibe a primeira opção do menu.
print("1 - Procurar por nome")

# Exibe a segunda opção do menu.
print("2 - Procurar por número")

# Pede ao usuário que escolha uma opção.
opcao = input("Escolha uma opção: ")

# Se a opção for "1", pesquisa pelo nome do Pokémon.
if opcao == "1":

    # Solicita o nome e converte para letras minúsculas.
    pesquisa = input("Digite o nome do Pokémon: ").lower()

# Se a opção for "2", pesquisa pelo número.
elif opcao == "2":

    # Solicita o número do Pokémon.
    pesquisa = input("Digite o número do Pokémon: ")

# Caso o usuário digite outra opção.
else:

    # Exibe uma mensagem de erro.
    print("Opção inválida!")

    # Encerra o programa.
    exit()

# Chama a função buscar_pokemon passando o nome ou número informado.
dados = buscar_pokemon(pesquisa)

# Verifica se a função retornou dados.
if dados:

    # Exibe um título indicando que o Pokémon foi encontrado.
    print("\n=== Pokémon Encontrado ===")

    # Mostra o nome do Pokémon.
    print("Nome:", dados["name"].title())

    # Mostra o número da Pokédex.
    print("Número:", dados["id"])

    # Mostra a altura.
    print("Altura:", dados["height"])

    # Mostra o peso.
    print("Peso:", dados["weight"])

    # Exibe o título da seção de tipos.
    print("\nTipos:")

    # Percorre a lista de tipos do Pokémon.
    for tipo in dados["types"]:

        # Mostra cada tipo com a primeira letra maiúscula.
        print("-", tipo["type"]["name"].title())

    # Exibe o título da seção de habilidades.
    print("\nHabilidades:")

    # Percorre a lista de habilidades.
    for habilidade in dados["abilities"]:

        # Mostra cada habilidade.
        print("-", habilidade["ability"]["name"].title())

    # Exibe o título da foto.
    print("\nFoto:")

    # Mostra o link da imagem do Pokémon.
    print(dados["sprites"]["front_default"])

# Executa este bloco caso o Pokémon não exista.
else:

    # Exibe uma mensagem informando que o Pokémon não foi encontrado.
    print("Pokémon não encontrado.")
```

```python
import requests
from tkinter import *
from PIL import Image, ImageTk
from io import BytesIO

# ---------------- FUNÇÃO ---------------- #

def buscar_pokemon():
    pesquisa = entrada.get().lower()

    url = f"https://pokeapi.co/api/v2/pokemon/{pesquisa}"
    resposta = requests.get(url)

    if resposta.status_code != 200:
        nome.config(text="Pokémon não encontrado!")
        numero.config(text="")
        altura.config(text="")
        peso.config(text="")
        tipos.config(text="")
        habilidades.config(text="")
        imagem.config(image="")
        return

    dados = resposta.json()

    # Dados
    nome.config(text=f"Nome: {dados['name'].title()}")
    numero.config(text=f"Nº: {dados['id']}")
    altura.config(text=f"Altura: {dados['height']}")
    peso.config(text=f"Peso: {dados['weight']}")

    lista_tipos = ", ".join(
        tipo["type"]["name"].title()
        for tipo in dados["types"]
    )

    tipos.config(text=f"Tipos: {lista_tipos}")

    lista_habilidades = ", ".join(
        habilidade["ability"]["name"].title()
        for habilidade in dados["abilities"]
    )

    habilidades.config(text=f"Habilidades: {lista_habilidades}")

    # Imagem
    url_imagem = dados["sprites"]["front_default"]

    if url_imagem:
        imagem_bytes = requests.get(url_imagem).content
        img = Image.open(BytesIO(imagem_bytes))
        img = img.resize((180, 180))

        foto = ImageTk.PhotoImage(img)

        imagem.config(image=foto)
        imagem.image = foto


# ---------------- JANELA ---------------- #

janela = Tk()
janela.title("Pokédex")
janela.geometry("450x650")
janela.resizable(False, False)

titulo = Label(
    janela,
    text="POKÉDEX",
    font=("Arial", 22, "bold")
)
titulo.pack(pady=10)

entrada = Entry(
    janela,
    font=("Arial", 15),
    justify="center"
)
entrada.pack(pady=10)

botao = Button(
    janela,
    text="Buscar",
    font=("Arial", 12),
    command=buscar_pokemon
)
botao.pack(pady=10)

imagem = Label(janela)
imagem.pack(pady=10)

nome = Label(janela, font=("Arial", 14))
nome.pack()

numero = Label(janela, font=("Arial", 14))
numero.pack()

altura = Label(janela, font=("Arial", 14))
altura.pack()

peso = Label(janela, font=("Arial", 14))
peso.pack()

tipos = Label(
    janela,
    font=("Arial", 14),
    wraplength=400,
    justify="center"
)
tipos.pack(pady=5)

habilidades = Label(
    janela,
    font=("Arial", 14),
    wraplength=400,
    justify="center"
)
habilidades.pack(pady=5)

janela.mainloop()
```
